```python
import numpy as np
import gymnasium as gym
import PyFlyt.gym_envs
from PyFlyt.gym_envs import FlattenWaypointEnv
from stable_baselines3 import PPO

MODEL_PATHS = {
    0: "/content/ppo_quadx_waypoints_1M.zip",       # existing Stage 2 model
    1: "/content/ppo_quadx_waypoints_1M_seed1.zip",
    2: "/content/ppo_quadx_waypoints_1M_seed2.zip",
    3: "/content/ppo_quadx_waypoints_1M_seed3.zip",
    4: "/content/ppo_quadx_waypoints_1M_seed4.zip",
}
models = {k: PPO.load(v, device="cpu") for k, v in MODEL_PATHS.items()}
print("Loaded members:", list(models.keys()))
```

```python
EVAL_SEEDS = list(range(1000, 1020))  # same seeds as Stage 2 evaluation
PRE_FAILURE_WINDOW = 20  # fixed step count; sensitivity-checked later (10/20/30)

episode_records = []
step_records = []

for seed in EVAL_SEEDS:
    env = gym.make("PyFlyt/QuadX-Waypoints-v4")
    env = FlattenWaypointEnv(env)
    obs, info = env.reset(seed=seed)

    ep_steps = []
    step_idx = 0
    done = False

    while not done:
        actions = np.stack([
            models[k].predict(obs, deterministic=True)[0] for k in range(5)
        ])  # shape (5, action_dim)

        disagreement_per_dim = actions.std(axis=0)
        total_disagreement = float(disagreement_per_dim.mean())

        exec_action = actions[0]  # member 0 flies
        obs, reward, terminated, truncated, info = env.step(exec_action)
        done = terminated or truncated

        ep_steps.append({
            "seed": seed,
            "step": step_idx,
            "total_disagreement": total_disagreement,
            "disagreement_per_dim": disagreement_per_dim.tolist(),
        })
        step_idx += 1

    if info["collision"]:
        outcome = "collision"
    elif info["out_of_bounds"]:
        outcome = "out_of_bounds"
    elif info["env_complete"]:
        outcome = "success"
    else:
        outcome = "timeout"

    episode_records.append({
        "seed": seed,
        "outcome": outcome,
        "num_targets_reached": info["num_targets_reached"],
        "episode_length": step_idx,
    })

    n = len(ep_steps)
    is_failure = outcome in ("collision", "out_of_bounds")
    for i, s in enumerate(ep_steps):
        s["outcome"] = outcome
        s["pre_failure"] = is_failure and (i >= n - PRE_FAILURE_WINDOW)
        step_records.append(s)

    env.close()
    print(f"seed {seed}: {outcome}, targets={info['num_targets_reached']}, len={step_idx}")

print(f"\nDone: {len(episode_records)} episodes, {len(step_records)} steps logged")
```

```python
import pickle

EVAL_SEEDS = list(range(1000, 1150))  # 150 episodes
PRE_FAILURE_WINDOW = 20               # sensitivity-checked later (10/20/30)

episode_records = []
step_records = []

for ep_i, seed in enumerate(EVAL_SEEDS):
    env = FlattenWaypointEnv(gym.make("PyFlyt/QuadX-Waypoints-v4"))
    obs, info = env.reset(seed=seed)

    ep_steps = []
    step_idx = 0
    done = False

    while not done:
        actions = np.stack([
            models[s].predict(obs, deterministic=True)[0] for s in MEMBER_SEEDS
        ])  # shape (n_members, action_dim)

        disagreement_per_dim = actions.std(axis=0)

        ep_steps.append({
            "seed": seed,
            "step": step_idx,
            "total_disagreement": float(disagreement_per_dim.mean()),
            "disagreement_per_dim": disagreement_per_dim.tolist(),
            "obs": obs.copy(),
        })

        exec_action = actions[MEMBER_SEEDS.index(EXECUTOR)]
        obs, reward, terminated, truncated, info = env.step(exec_action)
        done = terminated or truncated
        step_idx += 1

    if info["collision"]:
        outcome = "collision"
    elif info["out_of_bounds"]:
        outcome = "out_of_bounds"
    elif info["env_complete"]:
        outcome = "success"
    else:
        outcome = "timeout"

    episode_records.append({
        "seed": seed,
        "outcome": outcome,
        "num_targets_reached": info["num_targets_reached"],
        "episode_length": step_idx,
    })

    n = len(ep_steps)
    for i, s_rec in enumerate(ep_steps):
        s_rec["outcome"] = outcome
        s_rec["pre_collision"] = (outcome == "collision") and (i >= n - PRE_FAILURE_WINDOW)
        s_rec["pre_failure"] = (outcome in ("collision", "out_of_bounds")) and (i >= n - PRE_FAILURE_WINDOW)
        step_records.append(s_rec)

    env.close()
    print(f"[{ep_i+1}/{len(EVAL_SEEDS)}] seed {seed}: {outcome}, "
          f"targets={info['num_targets_reached']}, len={step_idx}")

# Persist immediately - session drop insurance
with open("/content/ensemble_eval_results.pkl", "wb") as f:
    pickle.dump({"episodes": episode_records, "steps": step_records,
                 "executor": EXECUTOR, "members": MEMBER_SEEDS,
                 "pre_failure_window": PRE_FAILURE_WINDOW}, f)

from collections import Counter
print("\nOutcome distribution:", Counter(r["outcome"] for r in episode_records))
print(f"{len(episode_records)} episodes, {len(step_records)} steps logged")
```

```python
import numpy as np
from collections import defaultdict, Counter
from sklearn.metrics import roc_auc_score, average_precision_score

thrust = np.array([s["disagreement_per_dim"][3] for s in step_records])
y_coll = np.array([s["pre_collision"] for s in step_records])
y_fail = np.array([s["pre_failure"] for s in step_records])

# ---------- 1. Average Precision: the honest metric for rare events ----------
base_rate_coll = y_coll.mean()
print("=== Average Precision (rare-event metric) ===")
print(f"pre_collision base rate: {base_rate_coll*100:.3f}%")
print(f"AP  (pre_collision): {average_precision_score(y_coll, thrust):.4f}  "
      f"[random baseline = {base_rate_coll:.4f}]")
print(f"AP  (pre_failure)  : {average_precision_score(y_fail, thrust):.4f}  "
      f"[random baseline = {y_fail.mean():.4f}]")

# ---------- 2. Time-matched control: is the signal just episode phase? ----------
print("\n=== Episode-length confound check ===")
by_outcome_len = defaultdict(list)
for r in episode_records:
    by_outcome_len[r["outcome"]].append(r["episode_length"])
for k, v in by_outcome_len.items():
    print(f"  {k:<14} n={len(v):>3}  mean_len={np.mean(v):6.1f}")

pos_idx = [s["step"] for s in step_records if s["pre_collision"]]
print(f"\npre_collision step-index range: {min(pos_idx)} - {max(pos_idx)}, "
      f"mean {np.mean(pos_idx):.1f}")

# Build a control pool: steps from NON-failure episodes at the SAME step indices
pos_index_counts = Counter(pos_idx)
matched_scores, matched_labels = [], []
nonfail_by_index = defaultdict(list)
for s in step_records:
    if s["outcome"] in ("timeout", "success"):
        nonfail_by_index[s["step"]].append(s["disagreement_per_dim"][3])

for s in step_records:
    if s["pre_collision"]:
        matched_scores.append(s["disagreement_per_dim"][3])
        matched_labels.append(1)
for idx, cnt in pos_index_counts.items():
    pool = nonfail_by_index.get(idx, [])
    for v in pool:
        matched_scores.append(v)
        matched_labels.append(0)

matched_scores = np.array(matched_scores)
matched_labels = np.array(matched_labels)
print(f"\nmatched set: {matched_labels.sum()} positive, "
      f"{(1-matched_labels).sum()} negative (same step indices only)")
print(f"AUROC (time-matched, thrust dim): "
      f"{roc_auc_score(matched_labels, matched_scores):.3f}")
print(f"AUROC (unmatched, for comparison): {roc_auc_score(y_coll, thrust):.3f}")

# ---------- 3. Is thrust disagreement simply a function of step index? ----------
print("\n=== Thrust disagreement vs step index (non-failure episodes only) ===")
for lo in range(0, 300, 50):
    vals = [v for i in range(lo, lo+50) for v in nonfail_by_index.get(i, [])]
    if vals:
        print(f"  steps {lo:>3}-{lo+49:>3}: mean={np.mean(vals):.4f}  n={len(vals)}")
```

```python
import numpy as np
from collections import defaultdict
from sklearn.metrics import roc_auc_score, average_precision_score, precision_recall_curve

thrust = np.array([s["disagreement_per_dim"][3] for s in step_records])
y_coll = np.array([s["pre_collision"] for s in step_records])

# ---------- 1. Concrete operating points (for the README) ----------
print("=== Precision at recall operating points (step-level, pre_collision) ===")
prec, rec, thr = precision_recall_curve(y_coll, thrust)
for target_recall in [0.3, 0.5, 0.7, 0.9]:
    i = np.argmin(np.abs(rec - target_recall))
    fp_per_tp = (1 - prec[i]) / prec[i] if prec[i] > 0 else float("inf")
    print(f"  recall={rec[i]:.2f} -> precision={prec[i]*100:.2f}%  "
          f"({fp_per_tp:.0f} false alarms per true catch)")

# ---------- 2. Episode-level aggregation ----------
print("\n=== Episode-level: thrust disagreement vs outcome ===")
ep_thrust = defaultdict(list)
for s in step_records:
    ep_thrust[s["seed"]].append(s["disagreement_per_dim"][3])

rows = []
for r in episode_records:
    v = np.array(ep_thrust[r["seed"]])
    rows.append({
        "seed": r["seed"],
        "outcome": r["outcome"],
        "targets": r["num_targets_reached"],
        "mean": v.mean(),
        "p90": np.percentile(v, 90),
        "max": v.max(),
    })

for oc in ["timeout", "success", "collision", "out_of_bounds"]:
    sub = [r for r in rows if r["outcome"] == oc]
    if sub:
        print(f"  {oc:<14} n={len(sub):>3}  "
              f"mean={np.mean([r['mean'] for r in sub]):.4f}  "
              f"p90={np.mean([r['p90'] for r in sub]):.4f}  "
              f"max={np.mean([r['max'] for r in sub]):.4f}")

y_ep = np.array([r["outcome"] in ("collision", "out_of_bounds") for r in rows])
print(f"\nepisode-level base rate: {y_ep.mean()*100:.1f}%  ({y_ep.sum()}/{len(y_ep)})")
for stat in ["mean", "p90", "max"]:
    x = np.array([r[stat] for r in rows])
    print(f"  AUROC({stat:>4}) = {roc_auc_score(y_ep, x):.3f}   "
          f"AP = {average_precision_score(y_ep, x):.3f}  "
          f"[baseline {y_ep.mean():.3f}]")

# ---------- 3. Correlation with mission performance ----------
print("\n=== Does disagreement track mission success? ===")
targets = np.array([r["targets"] for r in rows], dtype=float)
for stat in ["mean", "p90"]:
    x = np.array([r[stat] for r in rows])
    print(f"  corr(targets_reached, {stat}) = {np.corrcoef(targets, x)[0,1]:+.3f}")

# ---------- 4. Rolling window at step level ----------
print("\n=== Rolling-window smoothing (step-level, pre_collision) ===")
by_seed = defaultdict(list)
for s in step_records:
    by_seed[s["seed"]].append(s)
for K in [1, 10, 25, 50]:
    sc, lb = [], []
    for seed, steps in by_seed.items():
        vals = np.array([x["disagreement_per_dim"][3] for x in steps])
        csum = np.concatenate([[0], np.cumsum(vals)])
        for i, x in enumerate(steps):
            lo = max(0, i - K + 1)
            sc.append((csum[i+1] - csum[lo]) / (i + 1 - lo))
            lb.append(x["pre_collision"])
    sc, lb = np.array(sc), np.array(lb)
    print(f"  K={K:>3}: AUROC={roc_auc_score(lb, sc):.3f}  "
          f"AP={average_precision_score(lb, sc):.4f}")
```

```python
import numpy as np
from collections import defaultdict

WARMUP = 50  # skip takeoff phase; justified by earlier step-index table
             # (steps 0-49 mean=0.174 vs ~0.155 afterwards)

def build_series(steps, episodes):
    """Return {seed: np.array of per-step thrust disagreement}, ordered by step."""
    by_seed = defaultdict(list)
    for s in steps:
        v = s["thrust_disagreement"] if "thrust_disagreement" in s else s["disagreement_per_dim"][3]
        by_seed[s["seed"]].append((s["step"], v))
    return {seed: np.array([v for _, v in sorted(p)]) for seed, p in by_seed.items()}

def running_mean(v, K):
    c = np.concatenate([[0.0], np.cumsum(v)])
    idx = np.arange(len(v))
    lo = np.maximum(0, idx - K + 1)
    return (c[idx + 1] - c[lo]) / (idx + 1 - lo)

def evaluate_halt(series, episodes, K, tau, warmup=WARMUP):
    """Simulate a halt trigger offline: first step where running mean exceeds tau."""
    n_fail = n_fail_caught = n_ok = n_ok_fired = 0
    leads = []
    for r in episodes:
        v = series[r["seed"]]
        rm = running_mean(v, K)
        fired_at = None
        for i in range(warmup, len(rm)):
            if rm[i] > tau:
                fired_at = i
                break
        is_fail = r["outcome"] in ("collision", "out_of_bounds")
        if is_fail:
            n_fail += 1
            if fired_at is not None:
                n_fail_caught += 1
                leads.append(len(v) - fired_at)   # steps of warning before failure
        else:
            n_ok += 1
            if fired_at is not None:
                n_ok_fired += 1
    total_fired = n_fail_caught + n_ok_fired
    return {
        "recall": n_fail_caught / n_fail if n_fail else 0.0,
        "false_alarm_rate": n_ok_fired / n_ok if n_ok else 0.0,
        "precision": n_fail_caught / total_fired if total_fired else 0.0,
        "median_lead": float(np.median(leads)) if leads else 0.0,
        "n_fired": total_fired,
    }

# ---------- Configuration search on EXPLORATION set only ----------
expl_series = build_series(step_records, episode_records)
all_vals = np.concatenate([running_mean(v, 25) for v in expl_series.values()])

print("=== Halt trigger sweep (EXPLORATION set, seeds 1000-1149) ===")
print(f"{'K':>4} {'tau':>7} {'recall':>8} {'false_al':>9} {'prec':>7} {'lead':>7} {'fired':>6}")
for K in [10, 25, 50]:
    for q in [80, 85, 90, 93, 95]:
        tau = float(np.percentile(all_vals, q))
        m = evaluate_halt(expl_series, episode_records, K, tau)
        print(f"{K:>4} {tau:>7.4f} {m['recall']:>8.2f} {m['false_alarm_rate']:>9.2f} "
              f"{m['precision']:>7.2f} {m['median_lead']:>7.1f} {m['n_fired']:>6}")
```

```python
def evaluate_checkpoint(series, episodes, C, tau, warmup=WARMUP):
    """Causal decision at a single checkpoint: mean over [warmup, C) only."""
    n_fail_after = n_caught = n_ok = n_ok_fired = 0
    n_fail_total = n_fail_before = 0
    for r in episodes:
        v = series[r["seed"]]
        is_fail = r["outcome"] in ("collision", "out_of_bounds")
        if is_fail:
            n_fail_total += 1
        if len(v) < C:                      # episode ended before the checkpoint
            if is_fail:
                n_fail_before += 1
            continue
        score = v[warmup:C].mean()
        fired = score > tau
        if is_fail:
            n_fail_after += 1
            n_caught += fired
        else:
            n_ok += 1
            n_ok_fired += fired
    total_fired = n_caught + n_ok_fired
    return {
        "recall": n_caught / n_fail_after if n_fail_after else 0.0,
        "false_alarm": n_ok_fired / n_ok if n_ok else 0.0,
        "precision": n_caught / total_fired if total_fired else 0.0,
        "coverage": n_fail_after / n_fail_total if n_fail_total else 0.0,
        "n_fired": total_fired,
    }

print("=== Causal checkpoint decision (EXPLORATION set, seeds 1000-1149) ===")
print(f"{'C':>5} {'tau':>7} {'recall':>8} {'false_al':>9} {'prec':>7} {'cover':>7} {'fired':>6}")
for C in [75, 100, 125, 150]:
    prefix_scores = np.array([
        expl_series[r["seed"]][WARMUP:C].mean()
        for r in episode_records if len(expl_series[r["seed"]]) >= C
    ])
    for q in [70, 80, 85, 90]:
        tau = float(np.percentile(prefix_scores, q))
        m = evaluate_checkpoint(expl_series, episode_records, C, tau)
        print(f"{C:>5} {tau:>7.4f} {m['recall']:>8.2f} {m['false_alarm']:>9.2f} "
              f"{m['precision']:>7.2f} {m['coverage']:>7.2f} {m['n_fired']:>6}")
```
